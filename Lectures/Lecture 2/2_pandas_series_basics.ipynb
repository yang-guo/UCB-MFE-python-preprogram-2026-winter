## Setup: Import the Libraries

In almost every data analysis notebook, you'll see these imports at the top. The naming conventions `pd` and `np` are universal - use them!

```python
import pandas as pd
import numpy as np

print(f"Pandas version: {pd.__version__}")
print(f"NumPy version: {np.__version__}")
```

## What is a Series?

A `pd.Series` is a **one-dimensional labeled array**. Let's break that down:
- **One-dimensional**: It's a single column of data
- **Labeled**: Each value has an index (label) associated with it
- **Array**: Under the hood, it's a numpy array

Let's create our first Series using market capitalization data for the top 10 US stocks.

```python
# Market caps in billions USD
market_caps = pd.Series([2_991, 2_741, 1_635, 1_493, 1_123, 816, 767, 760, 559, 552])
market_caps
```

Notice three things in the output:
1. **Left column (0-9)**: The index - automatically assigned as integers
2. **Right column**: The values - our market cap data
3. **dtype**: The data type - `int64` means 64-bit integers

But this data isn't very useful without knowing which stock is which! Let's add a name to make it clearer:

```python
market_caps.name = 'Market caps of top 10 stocks in billions USD'
market_caps
```

Better! But we still don't know which number corresponds to which stock. This is where **custom indices** come in.

## Series Anatomy: dtype and values

Before we customize our index, let's peek under the hood. A Series has two key properties:

```python
# The data type
print("Data type:", market_caps.dtype)
print("Type of dtype:", type(market_caps.dtype))
```

```python
# The underlying values (a numpy array!)
print("Values:", market_caps.values)
print("Type:", type(market_caps.values))
```

**Key insight**: A Series is essentially a numpy array with a fancy index on top. This means:
- All the speed of numpy operations
- Plus the convenience of labeled data
- Best of both worlds!

## Custom Indices: Making Data Meaningful

The real power of Series comes from custom indices. Instead of 0, 1, 2..., let's use stock tickers!

```python
# Set custom index
market_caps.index = ['AAPL', 'MSFT', 'GOOG', 'AMZN', 'NVDA', 'META', 'BRK-B', 'TSLA', 'LLY', 'V']
market_caps
```

Much better! But doing this step-by-step is tedious. Pandas gives us shortcuts:

### Method 1: From a Dictionary

The most intuitive way - keys become indices, values become... values!

```python
market_caps = pd.Series(
    {
        'AAPL': 2_991, 
        'MSFT': 2_741, 
        'GOOG': 1_635, 
        'AMZN': 1_493, 
        'NVDA': 1_123, 
        'META': 816, 
        'BRK-B': 767, 
        'TSLA': 760, 
        'LLY': 559, 
        'V': 552
    }, 
    name='Market caps of top 10 stocks in billions USD'
)
market_caps
```

### Method 2: From Lists

When you have separate data and label lists:

```python
market_caps = pd.Series(
    [2_991, 2_741, 1_635, 1_493, 1_123, 816, 767, 760, 559, 552],
    index=['AAPL', 'MSFT', 'GOOG', 'AMZN', 'NVDA', 'META', 'BRK-B', 'TSLA', 'LLY', 'V'],
    name='Market caps of top 10 stocks in billions USD'
)
market_caps
```

**Bottom line**: Series construction is intuitive and flexible. Use whichever method fits your data source!

## Accessing Data: Index vs Position

This is **THE MOST IMPORTANT CONCEPT** to understand in pandas. There are two ways to access data:
1. **By index/label**: Using the custom index we set
2. **By position**: Using the numeric position (0, 1, 2...)

Let's explore both.

### Accessing by Index (Label)

Use `[]` with the index label:

```python
# Get Microsoft's market cap
market_caps['MSFT']
```

```python
# Get NVIDIA's market cap
market_caps['NVDA']
```

### The Index Trap: A Common Mistake

**WARNING**: This is where beginners get confused. Watch what happens when we sort the Series:

```python
# Create a simple Series with default index
series = pd.Series([10, 2, 3])
print("Original Series:")
print(series)
print("\nType of index:", type(series.index))
```

```python
# Access "first" element - what do we get?
print("series[0] =", series[0])  # We expect 10
```

```python
# Access "third" element
print("series[2] =", series[2])  # We expect 3
```

```python
# Now sort it (inplace=True modifies the Series directly)
series.sort_values(inplace=True)
print("After sorting:")
print(series)
```

**Notice**: After sorting, the indices stayed with their values! The index `0` still points to the value `10`, even though it's now in the third position.

```python
# This is still index 0 (not position 0!)
print("series[0] after sorting =", series[0])  # Still 10!
```

### The Solution: `.iloc[]` for Position-Based Access

To access by **position** (not index), use `.iloc[]`:

```python
# Compare index access vs position access
print("Index access:")
print("series[0] =", series[0])  # Index 0 -> value 10

print("\nPosition access:")
print("series.iloc[0] =", series.iloc[0])  # Position 0 -> value 2 (first element)
```

```python
# Let's verify with our market caps
print("First by index (AAPL):")
print(market_caps['AAPL'])

print("\nFirst by position:")
print(market_caps.iloc[0])
```

**Best Practice**: 
- Use `series['label']` when you know the index/label
- Use `series.iloc[position]` when you want the nth element
- When in doubt, be explicit with `.iloc[]`!

## Selecting Multiple Elements

We can select multiple elements by passing a list of labels or positions:

```python
# Select MSFT and NVDA by label
market_caps[ ['MSFT', 'NVDA'] ]
```

```python
# Select 2nd and 6th stocks by position (1 and 5 in 0-indexed)
market_caps.iloc[ [1, 5] ]
```

## Slicing: Ranges of Data

Slicing works with both index and position, but **they behave differently**!

```python
# Slice by index: INCLUSIVE of end point
print("From NVDA to LLY (inclusive):")
market_caps['NVDA':'LLY']
```

```python
# Slice by position: EXCLUSIVE of end point (like Python lists)
print("First two stocks (positions 0 and 1):")
market_caps.iloc[0:2]
```

**Critical difference**:
- `series['start':'end']` - **includes** both start and end
- `series.iloc[start:end]` - **excludes** end (like Python slicing)

This trips up even experienced users!

## Boolean Filtering: The Data Analyst's Superpower

Here's where pandas really shines. We can filter data using conditions. This is how you'll spend most of your time in real analysis!

### Step 1: Create a Boolean Series

When we compare a Series to a value, we get back a Series of True/False:

```python
# Which stocks have market cap under $1 trillion?
is_under_trillion = market_caps < 1_000
print("Boolean mask:")
print(is_under_trillion)
print("\nData type:", is_under_trillion.dtype)
```

### Step 2: Use the Boolean Series to Filter

Now we can use this boolean Series to filter the original:

```python
# Get only stocks under $1 trillion
market_caps[ is_under_trillion ]
```

### Inline Filtering (More Common)

Usually, we skip the intermediate variable:

```python
# Same result, one line
market_caps[market_caps < 1_000]
```

### Combining Conditions

We can combine multiple conditions using:
- `&` for AND
- `|` for OR  
- `~` for NOT

**Important**: You need parentheses around each condition!

```python
# Stocks under $1T AND ticker length is 4 characters
condition = (market_caps < 1_000) & (market_caps.index.str.len() == 4)
condition
```

```python
# Apply the filter
market_caps[condition]
```

```python
# Or inline (same result)
market_caps[(market_caps < 1_000) & (market_caps.index.str.len() == 4)]
```

**Why parentheses?** Because `&` and `|` have the same precedence as comparison operators. Without parentheses, you'll get errors or wrong results!

```python
# More examples
print("Stocks over $2T OR under $600B:")
print(market_caps[(market_caps > 2_000) | (market_caps < 600)])

print("\nStocks NOT in the $1T-$2T range:")
print(market_caps[~((market_caps >= 1_000) & (market_caps <= 2_000))])
```

## Vectorized Operations: NumPy Speed Meets Pandas Convenience

Remember: Series are numpy arrays underneath. This means all operations are **vectorized** - they apply to every element at once, incredibly fast!

### Arithmetic Operations

```python
# Convert billions to actual dollars
market_caps_dollars = market_caps * 1_000_000_000
print("Market caps in dollars:")
print(market_caps_dollars)
```

```python
# Calculate market cap per billion in market cap (silly but demonstrates division)
print("\nDivide all by 100:")
print(market_caps / 100)

print("\nSquare all values:")
print(market_caps ** 2)
```

### Aggregation Functions

These reduce the Series to a single value:

```python
# Average market cap
print(f"Mean: ${market_caps.mean():.2f}B")
```

```python
# Total market cap
print(f"Total: ${market_caps.sum():.0f}B")
```

```python
# Standard deviation
print(f"Std Dev: ${market_caps.std():.2f}B")
```

```python
# Other useful aggregations
print(f"Median: ${market_caps.median():.2f}B")
print(f"Min: ${market_caps.min():.0f}B")
print(f"Max: ${market_caps.max():.0f}B")
print(f"Range: ${market_caps.max() - market_caps.min():.0f}B")
```

### NumPy Functions Work Too!

Since Series are numpy arrays underneath, all numpy functions work:

```python
# Natural logarithm (useful for log returns!)
print("Log of market caps:")
print(np.log(market_caps))
```

```python
# NumPy sum (same as pandas sum)
print(f"\nNumPy sum: ${np.sum(market_caps):.0f}B")
print(f"Pandas sum: ${market_caps.sum():.0f}B")
print("They're the same!")
```

```python
# More numpy functions
print("Square root:")
print(np.sqrt(market_caps))

print("\nExponential:")
print(np.exp(market_caps / 1000))  # Scaled down to prevent overflow
```

## Modifying Series: In-Place Operations

Most operations create new Series (non-destructive). But sometimes we want to modify the original Series directly.

```python
# Create a fresh copy to modify
market_caps_copy = market_caps.copy()
print("Original:")
print(market_caps_copy)
```

### Update a Single Value

```python
# Tesla's market cap changed!
market_caps_copy[ 'TSLA' ] = 1000
print("After updating TSLA:")
print(market_caps_copy)
```

### Update by Position

```python
# Update 4th position (AMZN) - maybe there was an error
market_caps_copy.iloc[3] = 1500
print("After updating position 3 (AMZN):")
print(market_caps_copy)
```

### Bulk Updates with Boolean Indexing

This is powerful - update multiple values at once based on a condition:

```python
# Set all stocks under $1T to 0 (imagine filtering out small caps)
market_caps_copy[market_caps_copy < 1_000] = 0
print("After zeroing out sub-$1T stocks:")
print(market_caps_copy)
```

## Summary: What We've Learned

Let's recap the key concepts:

### 1. Series Structure
- A Series is a labeled, one-dimensional array
- It's a numpy array with an index on top
- Has a dtype (data type) and optional name

### 2. Creating Series
- From lists: `pd.Series([values], index=[labels])`
- From dicts: `pd.Series({label: value})`
- Both are equally valid!

### 3. The Index/Position Distinction
- `series['label']` - access by index/label
- `series.iloc[position]` - access by numeric position
- This distinction becomes critical after sorting!

### 4. Boolean Filtering
- Create boolean mask: `series > value`
- Apply mask: `series[series > value]`
- Combine conditions with `&`, `|`, `~` (with parentheses!)

### 5. Vectorized Operations
- All arithmetic operations work element-wise
- Aggregations: `.mean()`, `.sum()`, `.std()`, etc.
- All numpy functions work with Series

### 6. Modifying Series
- Update by label: `series['label'] = value`
- Update by position: `series.iloc[pos] = value`
- Bulk update: `series[condition] = value`

## Quick Practice (Follow Along)

Let's apply what we learned with a new dataset - daily returns for a stock:

```python
# Simulated daily returns (in %)
import pandas as pd
import numpy as np

np.random.seed(42)
returns = pd.Series(
    np.random.randn(10) * 2,  # 2% daily volatility
    index=pd.date_range('2024-01-01', periods=10),
    name='Daily Returns (%)'
)
returns
```

```python
# Find days with positive returns
print("Positive return days:")
print(returns[returns > 0])
```

```python
# Calculate statistics
print(f"Mean daily return: {returns.mean():.3f}%")
print(f"Volatility (std): {returns.std():.3f}%")
print(f"Best day: {returns.max():.3f}%")
print(f"Worst day: {returns.min():.3f}%")
```

```python
# Cumulative returns (this is what compounding looks like!)
cumulative = (1 + returns/100).cumprod() - 1
print("Cumulative returns:")
print(cumulative * 100)  # Back to percentage
```
